## Predictive Modelling

To reduce temporal leakage and better represent real-world use, the dataset is divided chronologically rather than using a random train-test split. Seasons 2012–2021 are used for training, 2022–2023 for validation and model selection, and 2024 is retained as the final out-of-time test set.

```python
from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

dummy_model = DummyRegressor(strategy="median")

dummy_model.fit(X_train, y_train)

dummy_val_pred_log = dummy_model.predict(X_val)

# Convert predictions and actual values back to euros
dummy_val_pred = np.expm1(dummy_val_pred_log)
y_val_eur = np.expm1(y_val)

dummy_mae = mean_absolute_error(
    y_val_eur,
    dummy_val_pred
)

dummy_rmse = np.sqrt(
    mean_squared_error(
        y_val_eur,
        dummy_val_pred
    )
)

dummy_r2 = r2_score(
    y_val_eur,
    dummy_val_pred
)

print(f"Naive Benchmark MAE: €{dummy_mae:,.0f}")
print(f"Naive Benchmark RMSE: €{dummy_rmse:,.0f}")
print(f"Naive Benchmark R²: {dummy_r2:.3f}")
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
import numpy as np
import pandas as pd

# ----------------------------
# Feature groups
# ----------------------------

categorical_features = [
    "competition_id",
    "country_of_citizenship",
    "position",
    "sub_position",
    "foot"
]

baseline_numeric = [
    "age",
    "goals",
    "assists",
    "minutes_played",
    "appearances",
    "height_in_cm",
    "yellow_cards",
    "red_cards"
]

enhanced_numeric = baseline_numeric + advanced_features


# ----------------------------
# Preprocessing function
# ----------------------------

def make_preprocessor(numeric_features):

    numeric_transformer = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ])

    categorical_transformer = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ])

    return ColumnTransformer([
        (
            "num",
            numeric_transformer,
            numeric_features
        ),
        (
            "cat",
            categorical_transformer,
            categorical_features
        )
    ])


# ----------------------------
# Same GB configuration
# ----------------------------

def make_gb_model(numeric_features):

    return Pipeline([
        (
            "preprocessor",
            make_preprocessor(numeric_features)
        ),
        (
            "model",
            GradientBoostingRegressor(
                n_estimators=400,
                learning_rate=0.05,
                max_depth=3,
                min_samples_leaf=10,
                subsample=1.0,
                random_state=42
            )
        )
    ])


baseline_gb = make_gb_model(baseline_numeric)
enhanced_gb = make_gb_model(enhanced_numeric)


# ----------------------------
# Train
# ----------------------------

baseline_gb.fit(
    X_train_base,
    y_train
)

enhanced_gb.fit(
    X_train_enh,
    y_train
)

print("Both models trained successfully.")
```

```python
def evaluate_model(model, X, y):

    pred_log = model.predict(X)

    actual_eur = np.expm1(y)
    predicted_eur = np.expm1(pred_log)

    return {
        "MAE_EUR":
            mean_absolute_error(actual_eur, predicted_eur),

        "RMSE_EUR":
            np.sqrt(
                mean_squared_error(actual_eur, predicted_eur)
            ),

        "Euro_R2":
            r2_score(actual_eur, predicted_eur),

        "Log_RMSE":
            np.sqrt(
                mean_squared_error(y, pred_log)
            ),

        "Log_R2":
            r2_score(y, pred_log)
    }


baseline_val_results = evaluate_model(
    baseline_gb,
    X_val_base,
    y_val
)

enhanced_val_results = evaluate_model(
    enhanced_gb,
    X_val_enh,
    y_val
)

comparison = pd.DataFrame(
    [
        baseline_val_results,
        enhanced_val_results
    ],
    index=[
        "Baseline",
        "Enhanced + Understat"
    ]
)

display(comparison)
```

```python
# ============================================================
# FINAL ENHANCED MODEL — TRAIN ON ALL 2014–2024 DATA
# ============================================================

from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.ensemble import GradientBoostingRegressor

# Same feature definitions used in evaluation
numeric_features_final = [
    "age",
    "goals",
    "assists",
    "minutes_played",
    "appearances",
    "height_in_cm",
    "yellow_cards",
    "red_cards",
    "xG_per90",
    "xA_per90",
    "npxG_per90",
    "shots_per90",
    "key_passes_per90",
    "xGChain_per90",
    "xGBuildup_per90"
]

categorical_features_final = [
    "competition_id",
    "country_of_citizenship",
    "position",
    "sub_position",
    "foot"
]

final_features = (
    numeric_features_final +
    categorical_features_final
)

# All historical enhanced observations
X_final_train = enhanced_model_df[final_features].copy()
y_final_train = enhanced_model_df["log_market_value"].copy()

# 2025/26 scoring data
X_2025_final = enhanced_2025_matched[final_features].copy()

# Preprocessing
numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ))
])

final_preprocessor = ColumnTransformer([
    ("num", numeric_transformer, numeric_features_final),
    ("cat", categorical_transformer, categorical_features_final)
])

# EXACT SAME locked Gradient Boosting specification
final_enhanced_model = Pipeline([
    ("preprocessor", final_preprocessor),
    ("model", GradientBoostingRegressor(
        n_estimators=400,
        learning_rate=0.05,
        max_depth=3,
        min_samples_leaf=10,
        subsample=1.0,
        random_state=42
    ))
])

# Train using ALL 2014–2024 observations
final_enhanced_model.fit(
    X_final_train,
    y_final_train
)

# Predict 2025/26 market values
pred_2025_log = final_enhanced_model.predict(
    X_2025_final
)

pred_2025_eur = np.expm1(pred_2025_log)

enhanced_2025_matched[
    "predicted_market_value_eur"
] = pred_2025_eur

print("Historical training observations:",
      len(X_final_train))

print("2025/26 predictions:",
      len(pred_2025_eur))

print("\nPredicted market value summary (€):")
print(
    enhanced_2025_matched[
        "predicted_market_value_eur"
    ].describe()
)
```

```python
# ============================================================
# ENHANCED MODEL 2024 TEST ERROR BY MARKET VALUE BAND
# ============================================================

test_error_bands = position_results.copy()

# We already have actual_eur and enhanced_pred_eur
test_error_bands["enhanced_abs_error"] = (
    test_error_bands["actual_eur"]
    - test_error_bands["enhanced_pred_eur"]
).abs()

# Define value bands
bins = [
    0,
    5_000_000,
    10_000_000,
    25_000_000,
    50_000_000,
    np.inf
]

labels = [
    "≤€5m",
    "€5–10m",
    "€10–25m",
    "€25–50m",
    ">€50m"
]

test_error_bands["value_band"] = pd.cut(
    test_error_bands["actual_eur"],
    bins=bins,
    labels=labels,
    include_lowest=True
)

enhanced_band_mae = (
    test_error_bands
    .groupby(
        "value_band",
        observed=False
    )
    .agg(
        observations=("actual_eur", "size"),
        MAE_EUR=("enhanced_abs_error", "mean")
    )
    .reset_index()
)

enhanced_band_mae["MAE_millions"] = (
    enhanced_band_mae["MAE_EUR"] / 1_000_000
)

display(
    enhanced_band_mae[
        [
            "value_band",
            "observations",
            "MAE_millions"
        ]
    ].round(2)
)
```

```python
# ============================================================
# FINAL TRANSFER RECOMMENDATION SCREEN
# ============================================================

deployment = deployment_2025.copy()

# Remove missing observed valuations
deployment = deployment[
    deployment["observed_market_value_eur"].notna()
].copy()

# ------------------------------------------------------------
# 1. Valuation gap
# ------------------------------------------------------------

deployment["valuation_gap_eur"] = (
    deployment["predicted_market_value_eur"]
    - deployment["observed_market_value_eur"]
)

deployment["valuation_gap_pct"] = (
    deployment["valuation_gap_eur"]
    / deployment["observed_market_value_eur"]
    * 100
)

# ------------------------------------------------------------
# 2. Assign observed market-value bands
# ------------------------------------------------------------

bins = [
    0,
    5_000_000,
    10_000_000,
    25_000_000,
    50_000_000,
    np.inf
]

labels = [
    "≤€5m",
    "€5–10m",
    "€10–25m",
    "€25–50m",
    ">€50m"
]

deployment["value_band"] = pd.cut(
    deployment["observed_market_value_eur"],
    bins=bins,
    labels=labels,
    include_lowest=True
)

# ------------------------------------------------------------
# 3. Map 2024 test MAE to each valuation band
# ------------------------------------------------------------

band_mae_map = {
    "≤€5m": 1.80e6,
    "€5–10m": 3.09e6,
    "€10–25m": 7.03e6,
    "€25–50m": 13.99e6,
    ">€50m": 35.29e6
}

deployment["band_mae_eur"] = (
    deployment["value_band"]
    .astype(str)
    .map(band_mae_map)
)

# Gap relative to normal model error
deployment["gap_to_mae_ratio"] = (
    deployment["valuation_gap_eur"]
    / deployment["band_mae_eur"]
)

# ------------------------------------------------------------
# 4. Recency of observed Transfermarkt valuation
# ------------------------------------------------------------

reference_date = pd.Timestamp("2026-06-12")

deployment["valuation_age_days"] = (
    reference_date
    - pd.to_datetime(deployment["observed_value_date"])
).dt.days

# ------------------------------------------------------------
# 5. Candidate screening
# ------------------------------------------------------------

candidate_pool = deployment[
    # Focus on recruitment-relevant price range
    (deployment["observed_market_value_eur"] <= 25_000_000)

    # Model predicts player above observed value
    & (deployment["valuation_gap_eur"] > 0)

    # Latest valuation reasonably recent
    & (deployment["valuation_age_days"] <= 30)
].copy()

# Rank primarily by gap relative to model's normal error
candidate_pool = candidate_pool.sort_values(
    "gap_to_mae_ratio",
    ascending=False
)

print("Eligible candidate pool:",
      len(candidate_pool))

# ------------------------------------------------------------
# 6. Final top 15
# ------------------------------------------------------------

shortlist_columns = [
    "player_id",
    "name",
    "age",
    "competition_id",
    "position",
    "sub_position",
    "observed_market_value_eur",
    "predicted_market_value_eur",
    "valuation_gap_eur",
    "valuation_gap_pct",
    "gap_to_mae_ratio",
    "observed_value_date"
]

final_shortlist = (
    candidate_pool[shortlist_columns]
    .head(15)
    .copy()
)

# Convert euro values to millions for readability
final_shortlist["Observed_Value_€m"] = (
    final_shortlist["observed_market_value_eur"]
    / 1_000_000
)

final_shortlist["Predicted_Value_€m"] = (
    final_shortlist["predicted_market_value_eur"]
    / 1_000_000
)

final_shortlist["Valuation_Gap_€m"] = (
    final_shortlist["valuation_gap_eur"]
    / 1_000_000
)

final_display = final_shortlist[
    [
        "name",
        "age",
        "competition_id",
        "position",
        "sub_position",
        "Observed_Value_€m",
        "Predicted_Value_€m",
        "Valuation_Gap_€m",
        "valuation_gap_pct",
        "gap_to_mae_ratio"
    ]
].copy()

final_display = final_display.rename(
    columns={
        "name": "Player",
        "age": "Age",
        "competition_id": "League",
        "position": "Position",
        "sub_position": "Sub-position",
        "valuation_gap_pct": "Gap_%",
        "gap_to_mae_ratio": "Gap_to_MAE"
    }
)

display(final_display.round(2))
```
